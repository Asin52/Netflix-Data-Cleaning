```python
missing_percentage=(df.isnull().sum()/len(df))*100
print(missing_percentage)
```

```python
missing_values=df.isnull().sum()
missing_values=missing_values[missing_values>0]
print(missing_values)
```

```python
df['director']=df['director'].fillna('unknown')
```

```python
df['director'].isnull().sum()
```

```python
df['cast']=df['cast'].fillna('unknown')
df['cast'].isnull().sum()
```

```python
df['country']=df['country'].fillna('unknown')
df['country'].isnull().sum()
```

```python
df['rating']=df['rating'].fillna('unknown')
df['rating'].isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df['show_id'].duplicated().sum()
```

```python
df[df.duplicated(subset=['show_id'], keep=False)]
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
text_columns = [
    'title',
    'director',
    'cast',
    'country',
    'rating',
    'listed_in',
    'description'
]

for column in text_columns:
    df[column] = df[column].str.strip()
```

```python
df['title'].head()
```

```python
df['type'] = df['type'].str.strip().str.title()

df['type'].unique()
```

```python
df['date_added'] = pd.to_datetime(
    df['date_added'],
    format='%B %d, %Y',
    errors='coerce'
)
```

```python
df['date_added'].dtype
```

```python
df['year_added'] = df['date_added'].dt.year

df[['date_added', 'year_added']].head()
```

```python
df['month_added'] = df['date_added'].dt.month

df[['date_added', 'month_added']].head()
```

```python
duration_split = df['duration'].str.extract(
    r'(\d+)\s*(min|seasons?)',
    flags=0
)

df['duration_value'] = pd.to_numeric(
    duration_split[0],
    errors='coerce'
)

df['duration_unit'] = duration_split[1].str.lower()
```

```python
df[['type', 'duration', 'duration_value', 'duration_unit']].head(10)
```

```python
df['release_year'] = pd.to_numeric(
    df['release_year'],
    errors='coerce'
).astype('Int64')
```

```python
df['release_year'].dtype
```

```python
df.isnull().sum()
```
